### ANOVA

```python
from scipy import stats


grupos = [grupo["felicidad"].values 
     for name, grupo in df.groupby("generacion")]


f_stat, p_value = stats.f_oneway(*grupos)

print("F =", f_stat)
print("p-value =", p_value)
```

Se realizó un ANOVA de un factor para analizar las diferencias en felicidad entre generaciones. Los resultados mostraron diferencias estadísticamente significativas, F(4, N−5) = 103.36, p < .001. Por lo tanto, se rechaza la hipótesis nula de igualdad de medias y se concluye que existen diferencias generacionales en los niveles de felicidad.

```python
from statsmodels.stats.multicomp import pairwise_tukeyhsd

tukey = pairwise_tukeyhsd(
    endog=df["felicidad"],   # o satisfaccion_vida
    groups=df["generacion"],
    alpha=0.05
)

print(tukey)
```

El análisis post-hoc de Tukey mostró diferencias significativas entre todas las generaciones (p < .05). La Generación Z presenta los niveles más altos de felicidad, seguida por Millennials y Generación X. No obstante, la Silent Generation muestra los niveles más bajos. La mayor diferencia se observa entre Generación Z y Silent Generation (0.60), lo que indica una brecha generacional bastante relevante en los niveles de bienestar.
