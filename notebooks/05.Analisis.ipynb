Esta variable no nos proporciona mucha información ya que muchos decidieron no contestar esta pregunta y es muy difícil sacar alguna tendencia de aquí.

```python
numeric_cols = df.select_dtypes(include="number").columns
```

```python
matrix = df[numeric_cols]
corr_matrix = matrix.corr()
corr_matrix
```

```python
plt.figure(figsize=(12,10))
plt.imshow(corr_matrix, cmap="cool")
plt.colorbar()
plt.xticks(range(len(corr_matrix.columns)), corr_matrix.columns, rotation=90)
plt.yticks(range(len(corr_matrix.columns)), corr_matrix.columns)
plt.title("Matriz de Correlación")
plt.show()
```
